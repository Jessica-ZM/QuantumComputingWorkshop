```python
# Puertas
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()

qc = QuantumCircuit(1)
qc.h(0)
qc.measure_all()

resultado = simulator.run(
    transpile(qc, simulator),
    shots=1000
).result()

cuentas = resultado.get_counts()
print(cuentas)

# Guardamos la figura devuelta por plot_histogram y la mostramos
fig = plot_histogram(cuentas)
display(fig)
```
