```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

simulator = AerSimulator()

qc = QuantumCircuit(2)

qc.x(0)
qc.cx(0, 1)

qc.measure_all()

print(qc.draw())

resultado = simulator.run(
    transpile(qc, simulator),
    shots=1000
).result()

cuentas = resultado.get_counts()

print(cuentas)
```
